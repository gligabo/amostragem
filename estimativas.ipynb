```python
ordem = [
    'LUIZ INÁCIO LULA DA SILVA',
    'JAIR MESSIAS BOLSONARO',
    'SIMONE NASSAR TEBET',
    'CIRO FERREIRA GOMES',
    'VOTO NULO',
    'VOTO BRANCO',
    'SORAYA VIEIRA THRONICKE',
    'LUIZ FELIPE CHAVES D AVILA',
    'KELMON LUIS DA SILVA SOUZA',
    'LEONARDO PÉRICLES VIEIRA ROQUE',
    'SOFIA PADUA MANZANO',
    'VERA LUCIA PEREIRA DA SILVA SALGADO',
    'JOSE MARIA EYMAEL'
]

def gerar_tabela_proporcoes_latex(
    estado_sigla: str,
    df_completo: pd.DataFrame,
    df_amostra: pd.DataFrame,
    nome_coluna_estado: str = 'estado'
) -> str:
    """
    Gera uma tabela LaTeX comparando proporções estimadas (da amostra)
    e verdadeiras (do DataFrame completo) para um dado estado.

    Args:
        estado_sigla (str): A sigla do estado (ex: 'AC', 'RO').
        df_completo (pd.DataFrame): O DataFrame completo do turno 1.
        df_amostra (pd.DataFrame): O DataFrame da amostra do estado.
        nome_coluna_estado (str): Nome da coluna que indica o estado (e.g., 'estado' ou 'Estado').

    Returns:
        str: A string contendo a tabela LaTeX.
    """

    if 'nome' in df_completo.columns:
        coluna_candidato = 'nome'
    elif 'Candidato' in df_completo.columns:
        coluna_candidato = 'Candidato'
    elif 'Presidente' in df_completo.columns:
        coluna_candidato = 'Presidente'
    else:
        raise ValueError("Coluna de candidato/voto não encontrada. Esperado 'nome', 'Candidato' ou 'Presidente'.")

    prop_est = (
        df_amostra[coluna_candidato]
        .value_counts(normalize = True)
        .reindex(ordem).fillna(0)
        .reset_index(drop = True)
    )

    prop_real = (
        df_completo[ df_completo[nome_coluna_estado] == estado_sigla ][coluna_candidato]
        .value_counts(normalize = True)
        .reindex(ordem)
        .fillna(0)
        .reset_index(drop = True)
    )

    data_prop_comparacao = pd.DataFrame( {
        'Candidato/Voto': ordem,
        'Proporção Estimada': prop_est,
        'Proporção Verdadeira': prop_real
    } )

    caption_text = f'Proporções Estimada vs. Verdadeira — Presidente {estado_sigla.upper()} 1 Turno'
    label_text = f'tab:proporcoes_{estado_sigla.lower()}_comparacao'

    latex_table = data_prop_comparacao.to_latex(
        index = False,
        float_format = '%.4f',
        caption = caption_text,
        label = label_text
    )
    return latex_table

# AC
latex_ac = gerar_tabela_proporcoes_latex(
    'AC', data_turno1_AC_RO, sample_1_AC, nome_coluna_estado = 'estado'
)
print("--- Tabela para AC ---")
print(latex_ac)


# RO
latex_ro = gerar_tabela_proporcoes_latex(
    'RO', data_turno1_AC_RO, sample_1_RO, nome_coluna_estado = 'estado'
)
print("\n--- Tabela para RO ---")
print(latex_ro)

# AM
latex_am = gerar_tabela_proporcoes_latex(
    'AM', data_turno1_AM_RR, sample_1_AM, nome_coluna_estado = 'Estado'
)
print("\n--- Tabela para AM ---")
print(latex_am)

# RR
latex_rr = gerar_tabela_proporcoes_latex(
    'RR', data_turno1_AM_RR, sample_1_RR, nome_coluna_estado = 'Estado'
)
print("\n--- Tabela para RR ---")
print(latex_rr)

# AP
latex_ap = gerar_tabela_proporcoes_latex(
    'AP', data_turno1_AP_PA_TO, sample_1_AP, nome_coluna_estado = 'Estado'
)
print("\n--- Tabela para AP ---")
print(latex_ap)

# PA
latex_pa = gerar_tabela_proporcoes_latex(
    'PA', data_turno1_AP_PA_TO, sample_1_PA, nome_coluna_estado = 'Estado'
)
print("\n--- Tabela para PA ---")
print(latex_pa)

# TO
latex_to = gerar_tabela_proporcoes_latex(
    'TO', data_turno1_AP_PA_TO, sample_1_TO, nome_coluna_estado = 'Estado'
)
print("\n--- Tabela para TO ---")
print(latex_to)
```

```python
N_h_values = {}

N_h_values['AC'] = len(data_turno1_AC_RO[ data_turno1_AC_RO['estado'] == 'AC' ])
N_h_values['RO'] = len(data_turno1_AC_RO[ data_turno1_AC_RO['estado'] == 'RO' ])

N_h_values['AM'] = len(data_turno1_AM_RR[ data_turno1_AM_RR['Estado'] == 'AM' ])
N_h_values['RR'] = len(data_turno1_AM_RR[ data_turno1_AM_RR['Estado'] == 'RR' ])

N_h_values['AP'] = len(data_turno1_AP_PA_TO[ data_turno1_AP_PA_TO['Estado'] == 'AP' ])
N_h_values['PA'] = len(data_turno1_AP_PA_TO[ data_turno1_AP_PA_TO['Estado'] == 'PA' ])
N_h_values['TO'] = len(data_turno1_AP_PA_TO[ data_turno1_AP_PA_TO['Estado'] == 'TO' ])

N_total = sum(N_h_values.values())

W_h_values = {state: N_h / N_total for state, N_h in N_h_values.items()}

def get_estimated_proportions_ph(
    df_amostra: pd.DataFrame,
    df_completo_original: pd.DataFrame
) -> pd.Series:
    """
    Calcula as proporções estimadas de candidatos/votos de uma amostra de estado.
    """
    if 'nome' in df_completo_original.columns:
        coluna_candidato = 'nome'
    elif 'Candidato' in df_completo_original.columns:
        coluna_candidato = 'Candidato'
    elif 'Presidente' in df_completo_original.columns:
        coluna_candidato = 'Presidente'
    else:
        raise ValueError("Coluna de candidato/voto não encontrada. Esperado 'nome', 'Candidato' ou 'Presidente'.")

    prop_est = (
        df_amostra[coluna_candidato]
        .value_counts(normalize = True)
        .reindex(ordem).fillna(0)
    )
    return prop_est

ph_values = {}
ph_values['AC'] = get_estimated_proportions_ph(sample_1_AC, data_turno1_AC_RO)
ph_values['RO'] = get_estimated_proportions_ph(sample_1_RO, data_turno1_AC_RO)
ph_values['AM'] = get_estimated_proportions_ph(sample_1_AM, data_turno1_AM_RR)
ph_values['RR'] = get_estimated_proportions_ph(sample_1_RR, data_turno1_AM_RR)
ph_values['AP'] = get_estimated_proportions_ph(sample_1_AP, data_turno1_AP_PA_TO)
ph_values['PA'] = get_estimated_proportions_ph(sample_1_PA, data_turno1_AP_PA_TO)
ph_values['TO'] = get_estimated_proportions_ph(sample_1_TO, data_turno1_AP_PA_TO)

estimated_overall_proportions = {}

for candidato in ordem:
    p_es_c = 0
    for state_sigla in W_h_values.keys():
        # p_es = sum(W_h * p_h)
        p_es_c += W_h_values[state_sigla] * ph_values[state_sigla][candidato]
    estimated_overall_proportions[candidato] = p_es_c

nord_candidates_ac_ro = data_turno1_AC_RO[data_turno1_AC_RO['estado'].isin(['AC', 'RO'])]['nome']
nord_candidates_am_rr = data_turno1_AM_RR[data_turno1_AM_RR['Estado'].isin(['AM', 'RR'])]['Presidente']
nord_candidates_ap_pa_to = data_turno1_AP_PA_TO[data_turno1_AP_PA_TO['Estado'].isin(['AP', 'PA', 'TO'])]['Candidato']

all_nord_candidates = pd.concat([
    nord_candidates_ac_ro,
    nord_candidates_am_rr,
    nord_candidates_ap_pa_to
])

true_overall_proportions = (
    all_nord_candidates
    .value_counts(normalize = True)
    .reindex(ordem)
    .fillna(0)
)

df_pes = pd.DataFrame({
    'Candidato/Voto': list(estimated_overall_proportions.keys()),
    'Proporção Estimada Global': list(estimated_overall_proportions.values()),
    'Proporção Verdadeira Global': true_overall_proportions.values
})

latex_pes = df_pes.to_latex(
    index = False,
    float_format = '%.4f',
    caption = 'Proporções Estimadas vs. Verdadeiras Globais — Presidente 1 Turno (Amostragem Estratificada Proporcional)',
    label = 'tab:proporcoes_globais_comparacao'
)

print("--- Tabela de Proporções Estimadas e Verdadeiras Globais (p_es) ---")
print(latex_pes)
```
