```python
import pandas as pd

votos_primeiro_turno = {
    'LUIZ INÁCIO LULA DA SILVA': 57259504,
    'JAIR MESSIAS BOLSONARO': 51072345,
    'SIMONE NASSAR TEBET': 4915423,
    'CIRO FERREIRA GOMES': 3599287,
    'VOTO NULO': 3487874,
    'VOTO BRANCO': 1964779,
    'SORAYA VIEIRA THRONICKE': 600955,
    'LUIZ FELIPE CHAVES D AVILA': 559708,
    'KELMON LUIS DA SILVA SOUZA': 81129,
    'LEONARDO PÉRICLES VIEIRA ROQUE': 53519,
    'SOFIA PADUA MANZANO': 45620,
    'VERA LUCIA PEREIRA DA SILVA SALGADO': 25625,
    'JOSE MARIA EYMAEL': 16604
}

lista_de_votos = []
for candidato, contagem in votos_primeiro_turno.items():
    lista_de_votos.extend([candidato] * contagem)

df_final_primeiro_turno = pd.DataFrame(lista_de_votos, columns=['Presidente'])


df_final_primeiro_turno = df_final_primeiro_turno.sample(frac=1, random_state = 0 ).reset_index(drop=True)
df_final_primeiro_turno.shape
```

```python
ordem_turno1 = [
    'LUIZ INÁCIO LULA DA SILVA',
    'JAIR MESSIAS BOLSONARO',
    'SIMONE NASSAR TEBET',
    'CIRO FERREIRA GOMES',
    'VOTO NULO',
    'VOTO BRANCO',
    'SORAYA VIEIRA THRONICKE',
    'LUIZ FELIPE CHAVES D AVILA',
    'KELMON LUIS DA SILVA SOUZA',
    'LEONARDO PÉRICLES VIEIRA ROQUE',
    'SOFIA PADUA MANZANO',
    'VERA LUCIA PEREIRA DA SILVA SALGADO',
    'JOSE MARIA EYMAEL'
]
proporcao_real_t1 = (
    df_final_primeiro_turno['Presidente']
    .value_counts(normalize=True)
    .reindex(ordem_turno1)
    .fillna(0)
)

amostra_t1 = df_final_primeiro_turno.sample(n=2401, random_state=16)

proporcao_amostra_t1 = (
    amostra_t1['Presidente']
    .value_counts(normalize=True)
    .reindex(ordem_turno1)
    .fillna(0)
)

df_comparacao_t1 = pd.DataFrame({
    'Candidato/Voto': proporcao_real_t1.index,
    'Proporção Real': proporcao_real_t1.values,
    'Proporção na Amostra': proporcao_amostra_t1.values
})

latex_t1 = df_comparacao_t1.to_latex(
    index=False,
    float_format="%.4f",
    caption="Proporção Real vs. Amostra (n=2401) - Eleição 1º Turno",
    label="tab:comp_t1"
)

print(latex_t1)
```
